```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import *
from IPython.display import Audio
from scipy.io import wavfile
from scipy.signal import find_peaks
```

```python
import colorsys

def hsl_to_hex(h, s, l):
    # Convert HSL to RGB
    r, g, b = colorsys.hls_to_rgb(h / 360.0, l / 100.0, s / 100.0)

    # Convert RGB values to hexadecimal
    return f'#{int(r * 255):02x}{int(g * 255):02x}{int(b * 255):02x}'

```

```python
def frec_col(n):
  p = (n * 100) / 20000
  c = (p * 280) / 100
  return c
```

#Audio file

```python
## Read the audio file, fs:sampling rate, nota: signal

fs, nota = wavfile.read("/content/Audio_Volador_1.wav")
nota.shape
```

```python
a, b, c, d, e, f, g, h = np.split(nota, 8, axis=0)
```

##A
First section of the audio file

```python
n = len(a)
duration = n/fs
dt = duration/n
t = np.arange(0,duration,dt)
```

```python
Fw_c1 = fft(a)
Freq_c1 = fftfreq(n, dt)
normal = abs(Fw_c1)/np.max(abs(Fw_c1))
```

```python
plt.plot(Freq_c1, normal, color="#00a9ff")
plt.xlim(0,30000)
plt.ylim(0,1)
plt.xlabel('Frecuencia (Hz)')
plt.ylabel('Intensidad')
plt.show()
```

```python
peaks, _ = find_peaks(normal, height=0.1) # Input (signal, threshold)
```

```python
color = hsl_to_hex(frec_col(Freq_c1[peaks[0]]),100,50)
```

```python
plt.plot(Freq_c1, normal, color="black")

for i in range(len(peaks)):
  plt.plot(Freq_c1[peaks[i]],
           normal[peaks[i]],
           "y*",
           color = hsl_to_hex(frec_col(Freq_c1[peaks[i]]),100,50))

plt.xlim(0,20000)
plt.ylim(0.1,1)
plt.xlabel('Frecuencia (Hz)')
plt.ylabel('Intensidad')
plt.show()
```

```python
for i in range(len(peaks)):
  if Freq_c1[peaks[i]] > 0:
    print(round(Freq_c1[peaks[i]]),0)
```

```python
print('The fundamental frequency is %d Hz' %Freq_c1[peaks[0]])
```
